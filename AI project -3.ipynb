```python
import pyttsx3
import time

class RoboSpeaker:
    def __init__(self):
        self.engine = pyttsx3.init()
        self.setup_engine()
        
    def setup_engine(self):
        """Configure the TTS engine settings"""
        # Get available voices
        voices = self.engine.getProperty('voices')
        
        # Set properties
        self.engine.setProperty('rate', 150)  # Speed of speech
        self.engine.setProperty('volume', 1.0)  # Volume (0.0 to 1.0)
        
        # Try to set a more robotic voice if available
        for voice in voices:
            if "robotic" in voice.name.lower() or "zira" in voice.name.lower():
                self.engine.setProperty('voice', voice.id)
                break
    
    def speak(self, text):
        """Convert text to speech"""
        print(f"\nRobo Speaker: {text}")
        self.engine.say(text)
        self.engine.runAndWait()
    
    def run(self):
        """Main interaction loop"""
        print("\n" + "="*40)
        print("ROBO SPEAKER 3000".center(40))
        print("="*40)
        print("Type 'quit' to exit the program.\n")
        
        while True:
            text = input("Enter what you want me to speak: ")
            
            if text.lower() == 'quit':
                self.speak("Goodbye! Shutting down Robo Speaker.")
                print("\nRobo Speaker has been shut down.")
                break
                
            if text.strip():
                self.speak(text)
            else:
                print("Please enter some text or 'quit' to exit.")

if __name__ == "__main__":
    try:
        speaker = RoboSpeaker()
        speaker.run()
    except KeyboardInterrupt:
        print("\nProgram interrupted by user. Goodbye!")
    except Exception as e:
        print(f"\nAn error occurred: {e}")
```

```python
import speech_recognition as sr
import pyttsx3
import time
import os

class SpeechRecognitionApp:
    def __init__(self):
        self.recognizer = sr.Recognizer()
        self.microphone = sr.Microphone()
        self.engine = pyttsx3.init()
        self.setup_voice_engine()
        
    def setup_voice_engine(self):
        """Configure the TTS engine settings"""
        voices = self.engine.getProperty('voices')
        self.engine.setProperty('rate', 150)
        self.engine.setProperty('volume', 1.0)
        # Try to set a female voice if available
        for voice in voices:
            if "female" in voice.name.lower():
                self.engine.setProperty('voice', voice.id)
                break
    
    def speak(self, text):
        """Convert text to speech"""
        print(f"Assistant: {text}")
        self.engine.say(text)
        self.engine.runAndWait()
    
    def listen(self):
        """Listen to microphone input and return recognized text"""
        with self.microphone as source:
            print("\nListening... (speak now)")
            self.recognizer.adjust_for_ambient_noise(source)
            audio = self.recognizer.listen(source, timeout=5, phrase_time_limit=8)
            
        try:
            print("Processing your speech...")
            # Using Google Web Speech API
            text = self.recognizer.recognize_google(audio)
            print(f"You said: {text}")
            return text.lower()
        except sr.UnknownValueError:
            self.speak("Sorry, I didn't understand that.")
            return ""
        except sr.RequestError as e:
            self.speak(f"Could not request results from service; {e}")
            return ""
        except Exception as e:
            self.speak(f"An error occurred: {e}")
            return ""
    
    def process_command(self, command):
        """Process the recognized command"""
        if not command:
            return
        
        if "hello" in command:
            self.speak("Hello there! How can I help you?")
        elif "time" in command:
            current_time = time.strftime("%I:%M %p")
            self.speak(f"The current time is {current_time}")
        elif "date" in command:
            current_date = time.strftime("%B %d, %Y")
            self.speak(f"Today's date is {current_date}")
        elif "your name" in command:
            self.speak("I am your voice assistant. You can call me Robo.")
        elif "exit" in command or "quit" in command or "goodbye" in command:
            self.speak("Goodbye! Have a nice day.")
            return True
        else:
            self.speak("I heard you say: " + command)
            self.speak("Sorry, I'm not programmed to handle this command yet.")
    
    def run(self):
        """Main application loop"""
        self.speak("Initializing voice assistant. Please wait...")
        print("\n" + "="*50)
        print("SPEECH RECOGNITION ASSISTANT".center(50))
        print("="*50)
        print("\nAvailable commands:")
        print("- Say 'hello' to greet the assistant")
        print("- Ask 'what time is it' to get current time")
        print("- Ask 'what is today's date' to get the date")
        print("- Ask 'what is your name'")
        print("- Say 'exit', 'quit', or 'goodbye' to end the program\n")
        
        self.speak("Assistant ready. How can I help you?")
        
        while True:
            command = self.listen()
            if self.process_command(command):
                break

if __name__ == "__main__":
    try:
        app = SpeechRecognitionApp()
        app.run()
    except KeyboardInterrupt:
        print("\nProgram terminated by user.")
    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
import speech_recognition as sr
import pyttsx3
import datetime
import time

class VoiceAssistant:
    def __init__(self):
        self.recognizer = sr.Recognizer()
        self.microphone = sr.Microphone()
        self.engine = pyttsx3.init()
        self.setup_voice()
    
    def setup_voice(self):
        """Configure TTS voice settings"""
        voices = self.engine.getProperty('voices')
        self.engine.setProperty('rate', 170)  # Speech speed
        self.engine.setProperty('volume', 1.0)  # Volume (0.0 to 1.0)
        # Use a female voice if available
        for voice in voices:
            if "female" in voice.name.lower():
                self.engine.setProperty('voice', voice.id)
                break
    
    def speak(self, text):
        """Make the assistant speak"""
        print(f"Assistant: {text}")
        self.engine.say(text)
        self.engine.runAndWait()
    
    def listen(self):
        """Listen to microphone and return recognized text"""
        with self.microphone as source:
            print("\nListening... (Speak now)")
            self.recognizer.adjust_for_ambient_noise(source, duration=1)
            try:
                audio = self.recognizer.listen(source, timeout=5, phrase_time_limit=5)
                text = self.recognizer.recognize_google(audio).lower()
                print(f"You said: {text}")
                return text
            except sr.UnknownValueError:
                self.speak("Sorry, I didn't catch that.")
                return ""
            except sr.RequestError:
                self.speak("Sorry, my speech service is down.")
                return ""
            except Exception as e:
                self.speak(f"An error occurred: {e}")
                return ""
    
    def handle_command(self, command):
        """Process user commands"""
        if not command:
            return False
        
        if "hello" in command or "hi" in command:
            self.speak("Hello! How can I help you?")
        
        elif "time" in command:
            current_time = datetime.datetime.now().strftime("%I:%M %p")
            self.speak(f"The current time is {current_time}")
        
        elif "date" in command:
            current_date = datetime.datetime.now().strftime("%B %d, %Y")
            self.speak(f"Today's date is {current_date}")
        
        elif "your name" in command:
            self.speak("I'm your voice assistant. Call me Alex!")
        
        elif "exit" in command or "quit" in command or "goodbye" in command:
            self.speak("Goodbye! Have a great day.")
            return True  # Signal to exit
        
        else:
            self.speak("I didn't understand that. Try saying 'time', 'date', or 'hello'.")
        
        return False
    
    def run(self):
        """Main program loop"""
        self.speak("Initializing voice assistant... Ready!")
        print("\n" + "="*50)
        print("VOICE ASSISTANT ACTIVATED".center(50))
        print("="*50)
        print("\nTry saying:")
        print("- 'Hello' or 'Hi'")
        print("- 'What time is it?'")
        print("- 'What is today's date?'")
        print("- 'What is your name?'")
        print("- 'Exit' or 'Goodbye' to quit\n")
        
        while True:
            command = self.listen()
            should_exit = self.handle_command(command)
            if should_exit:
                break

if __name__ == "__main__":
    try:
        assistant = VoiceAssistant()
        assistant.run()
    except KeyboardInterrupt:
        print("\nProgram stopped by user.")
    except Exception as e:
        print(f"Error: {e}")
```
