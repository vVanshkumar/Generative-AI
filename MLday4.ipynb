```python
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris
import numpy as np
import matplotlib.pyplot as plt

# Load data
iris = load_iris()
X = iris["data"][:, 3:]  # Use only one feature
y = (iris["target"] == 2).astype(int)  # Binary classification for "virginica"

# Train a logistic regression model
clf = LogisticRegression()
clf.fit(X, y)

# Create new data for predictions
X_new = np.linspace(0, 3, 100).reshape(-1, 1)  # Example range for feature
y_prob = clf.predict_proba(X_new)  # Predict probabilities

# Plot the probabilities for the "virginica" class
plt.plot(X_new, y_prob[:, 1], "g-", label="virginica")  # Index `1` for positive class
plt.xlabel("Feature value")
plt.ylabel("Probability")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression

# Load iris dataset and extract features
iris = load_iris()
X = iris["data"][:, 3:]  # Petal width as the feature
y = (iris["target"] == 2).astype(int)  # Binary target for "virginica"

# Train logistic regression
clf = LogisticRegression()
clf.fit(X, y)

# Create new feature range for predictions
X_new = np.linspace(0, 3, 100).reshape(-1, 1)
y_prob = clf.predict_proba(X_new)  # Predict probabilities

# Plot the predicted probability for "virginica"
plt.plot(X_new, y_prob[:, 1], "g-", label="virginica")  # Column 1 corresponds to class "virginica"
plt.xlabel("Petal width")
plt.ylabel("Probability")
plt.title("Probability of being Virginica")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression

# Load iris dataset and extract features
iris = load_iris()
X = iris["data"][:, 3:]  # Petal width as the feature
y = (iris["target"] == 2).astype(int)  # Binary target for "virginica"

# Train logistic regression
clf = LogisticRegression()
clf.fit(X, y)

# Create new feature range for predictions
X_new = np.linspace(0, 3, 100).reshape(-1, 1)
y_prob = clf.predict_proba(X_new)  # Predict probabilities

# Plot the predicted probability for "virginica"
plt.plot(X_new, y_prob[:, 1], "g-", label="virginica")  

plt.xlabel("Petal Width")
plt.ylabel("Probability")


plt.title("Probability of being Virginica")
plt.legend()
plt.show()
```
