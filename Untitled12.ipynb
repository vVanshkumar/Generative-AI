```python

# =====================================================
# SIMPLE YOLO REAL-TIME OBJECT DETECTION PROJECT
# =====================================================

import cv2
from ultralytics import YOLO

# 1. Load YOLO model
print("Loading YOLO model...")

model = YOLO("yolov8n.pt")

print("YOLO model loaded successfully!")

# 2. Start webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("ERROR: Webcam not detected!")
else:

    print("Webcam started successfully!")
    print("Object detection is running...")
    print("Press Q to exit.")

    try:

        while True:

            # 3. Read webcam frame
            ret, frame = cap.read()

            if not ret:
                print("Cannot read webcam!")
                break

            # 4. Detect objects using YOLO
            results = model.predict(
                source=frame,
                conf=0.40,
                imgsz=640,
                verbose=False
            )

            # 5. Draw bounding boxes
            output = results[0].plot()

            # 6. Display project title
            cv2.putText(
                output,
                "YOLO REAL-TIME OBJECT DETECTION",
                (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 255),
                2
            )

            # 7. Display webcam output
            cv2.imshow(
                "YOLO Object Detection",
                output
            )

            # 8. Press Q to stop
            if cv2.waitKey(1) & 0xFF == ord('q'):
                break

    finally:

        # 9. Release webcam
        cap.release()

        # 10. Close windows
        cv2.destroyAllWindows()

        print("Detection stopped successfully!")

# =====================================================
# PROJECT COMPLETED
# =====================================================
```
