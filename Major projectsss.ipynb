```python
import speech_recognition as sr
from collections import defaultdict
import time

class SpeechRecognizer:
    def __init__(self):
        self.recognizer = sr.Recognizer()
        self.microphone = sr.Microphone()
        self.keyword_counts = defaultdict(int)
        self.keywords = ["python", "code", "program", "algorithm", "function"]
        
    def adjust_for_noise(self):
        with self.microphone as source:
            print("Adjusting for ambient noise...")
            self.recognizer.adjust_for_ambient_noise(source, duration=1)
    
    def listen_and_recognize(self, timeout=5, phrase_time_limit=10):
        try:
            with self.microphone as source:
                print(f"Listening for {timeout} seconds...")
                audio = self.recognizer.listen(
                    source, timeout=timeout, phrase_time_limit=phrase_time_limit)
                
                print("Processing speech...")
                text = self.recognizer.recognize_google(audio).lower()
                print(f"Recognized: {text}")
                
                # Check for keywords
                for word in self.keywords:
                    if word in text:
                        self.keyword_counts[word] += 1
                        print(f"Detected keyword: {word} (total: {self.keyword_counts[word]})")
                
                return text
                
        except sr.WaitTimeoutError:
            print("No speech detected within the timeout period.")
            return None
        except sr.UnknownValueError:
            print("Could not understand audio.")
            return None
        except sr.RequestError as e:
            print(f"Error with the speech recognition service: {e}")
            return None
    
    def continuous_listening(self, duration=60):
        self.adjust_for_noise()
        start_time = time.time()
        
        while time.time() - start_time < duration:
            self.listen_and_recognize()
            print("Keyword counts:", dict(self.keyword_counts))
            time.sleep(1)

if __name__ == "__main__":
    sr = SpeechRecognizer()
    sr.continuous_listening(duration=30)  # Listen for 30 seconds
```

```python
import speech_recognition as sr
import pyttsx3
from datetime import datetime

# Initialize text-to-speech engine
engine = pyttsx3.init()

def speak(text):
    """Convert text to speech"""
    engine.say(text)
    engine.runAndWait()

def get_audio():
    """Listen to microphone and return recognized text"""
    recognizer = sr.Recognizer()
    with sr.Microphone() as source:
        print("Listening...")
        recognizer.adjust_for_ambient_noise(source, duration=1)
        audio = recognizer.listen(source, timeout=5)
        
    try:
        text = recognizer.recognize_google(audio).lower()
        print(f"You said: {text}")
        return text
    except sr.UnknownValueError:
        speak("Sorry, I didn't understand that.")
    except sr.RequestError:
        speak("Sorry, my speech service is down.")
    return ""

def get_time():
    """Returns current time in a human-readable format"""
    now = datetime.now()
    current_time = now.strftime("%I:%M %p")  # Example: "03:45 PM"
    return f"The current time is {current_time}"

def get_date():
    """Returns current date in a human-readable format"""
    now = datetime.now()
    current_date = now.strftime("%A, %B %d, %Y")  # Example: "Monday, January 01, 2024"
    return f"Today is {current_date}"

def main():
    speak("Hello! How can I help you?")
    while True:
        command = get_audio()
        
        if not command:
            continue
        
        if "time" in command:
            time_response = get_time()
            print(time_response)
            speak(time_response)
        
        elif "date" in command:
            date_response = get_date()
            print(date_response)
            speak(date_response)
        
        elif "stop" in command or "exit" in command:
            speak("Goodbye!")
            break
        
        else:
            speak("I can tell you the time or date. Try saying 'What is the time?'")

if __name__ == "__main__":
    main()
```

```python
import speech_recognition as sr
import pyttsx3
import datetime
import time

# Initialize recognizer and text-to-speech engine
recognizer = sr.Recognizer()
engine = pyttsx3.init()

def speak(text):
    """Convert text to speech"""
    engine.say(text)
    engine.runAndWait()

def listen():
    """Listen to microphone and return recognized text"""
    with sr.Microphone() as source:
        print("Listening...")
        recognizer.adjust_for_ambient_noise(source, duration=0.5)
        audio = recognizer.listen(source)
        
        try:
            text = recognizer.recognize_google(audio).lower()
            print(f"You said: {text}")
            return text
        except sr.UnknownValueError:
            speak("Sorry, I didn't catch that.")
            return ""
        except sr.RequestError:
            speak("Sorry, my speech service is down.")
            return ""

def respond(command):
    """Respond to voice commands"""
    if "hello" in command or "hi" in command:
        speak("Hello! How can I help you?")
    elif "time" in command:
        current_time = datetime.datetime.now().strftime("%I:%M %p")
        speak(f"The current time is {current_time}")
    elif "date" in command:
        current_date = datetime.datetime.now().strftime("%B %d, %Y")
        speak(f"Today's date is {current_date}")
    elif "your name" in command:
        speak("I am a simple voice assistant.")
    elif "exit" in command or "bye" in command:
        speak("Goodbye! Have a nice day.")
        exit()
    else:
        speak("I didn't understand that command.")

def main():
    speak("Hello! I am your voice assistant. How can I help you?")
    while True:
        command = listen()
        if command:
            respond(command)
        time.sleep(1)  # Small delay to avoid CPU overload

if __name__ == "__main__":
    main()
```

```python
import speech_recognition as sr
import pyttsx3
import datetime
import time
import webbrowser
import os

class VoiceAssistant:
    def __init__(self):
        self.recognizer = sr.Recognizer()
        self.engine = pyttsx3.init()
        self.setup_voice()
        
    def setup_voice(self):
        """Configure voice settings"""
        voices = self.engine.getProperty('voices')
        self.engine.setProperty('voice', voices[0].id)  # 0 for male, 1 for female
        self.engine.setProperty('rate', 150)  # Speech speed

    def speak(self, text):
        """Convert text to speech"""
        print(f"Assistant: {text}")
        self.engine.say(text)
        self.engine.runAndWait()

    def listen(self):
        """Listen to microphone input"""
        with sr.Microphone() as source:
            print("\nListening...")
            self.recognizer.adjust_for_ambient_noise(source)
            audio = self.recognizer.listen(source, timeout=5)
            
            try:
                text = self.recognizer.recognize_google(audio).lower()
                print(f"You said: {text}")
                return text
            except sr.UnknownValueError:
                self.speak("Sorry, I didn't catch that.")
                return ""
            except sr.RequestError:
                self.speak("Sorry, my speech service is down.")
                return ""
            except Exception as e:
                self.speak(f"An error occurred: {e}")
                return ""

    def process_command(self, command):
        """Process voice commands"""
        if not command:
            return
            
        if "hello" in command or "hi" in command:
            self.speak("Hello! How can I help you today?")
            
        elif "time" in command:
            current_time = datetime.datetime.now().strftime("%I:%M %p")
            self.speak(f"The current time is {current_time}")
            
        elif "date" in command:
            current_date = datetime.datetime.now().strftime("%B %d, %Y")
            self.speak(f"Today's date is {current_date}")
            
        elif "search" in command:
            query = command.replace("search", "").strip()
            if query:
                url = f"https://www.google.com/search?q={query}"
                webbrowser.open(url)
                self.speak(f"Searching for {query} on Google")
            else:
                self.speak("What would you like me to search?")
                
        elif "open youtube" in command:
            webbrowser.open("https://youtube.com")
            self.speak("Opening YouTube")
            
        elif "open notepad" in command:
            os.system("notepad.exe")
            self.speak("Opening Notepad")
            
        elif "your name" in command:
            self.speak("I'm your personal voice assistant")
            
        elif "exit" in command or "bye" in command:
            self.speak("Goodbye! Have a nice day.")
            exit()
            
        else:
            self.speak("I didn't understand that command. Can you repeat?")

    def run(self):
        """Main execution loop"""
        self.speak("Voice Assistant activated. How can I help you?")
        
        while True:
            command = self.listen()
            self.process_command(command)
            time.sleep(1)  # Prevent CPU overload

if __name__ == "__main__":
    assistant = VoiceAssistant()
    assistant.run()
```

```python
import speech_recognition as sr
import pyttsx3
import webbrowser
import time
import sys
import os

class VoiceBrowserAssistant:
    def __init__(self):
        self.recognizer = sr.Recognizer()
        self.engine = pyttsx3.init()
        self.setup_voice()
        self.websites = {
            "youtube": "https://www.youtube.com",
            "facebook": "https://www.facebook.com",
            "instagram": "https://www.instagram.com",
            "google": "https://www.google.com",
            "twitter": "https://twitter.com",
            "linkedin": "https://linkedin.com"
        }
        self.running = True

    def setup_voice(self):
        """Configure voice settings"""
        voices = self.engine.getProperty('voices')
        self.engine.setProperty('voice', voices[1].id)  # Female voice
        self.engine.setProperty('rate', 150)  # Speech speed

    def speak(self, text):
        """Convert text to speech"""
        print(f"Assistant: {text}")
        self.engine.say(text)
        self.engine.runAndWait()

    def listen(self):
        """Listen to microphone input"""
        with sr.Microphone() as source:
            print("\nListening... (Try: 'open YouTube', 'open Google', or 'exit')")
            self.recognizer.adjust_for_ambient_noise(source)
            try:
                audio = self.recognizer.listen(source, timeout=5)
                command = self.recognizer.recognize_google(audio).lower()
                print(f"You said: {command}")
                return command
            except sr.WaitTimeoutError:
                self.speak("I didn't hear anything. Try again.")
                return ""
            except sr.UnknownValueError:
                self.speak("Sorry, I didn't understand that.")
                return ""
            except sr.RequestError:
                self.speak("Sorry, my speech service is down.")
                return ""
            except Exception as e:
                self.speak(f"An error occurred: {str(e)}")
                return ""

    def open_website(self, site_name):
        """Open a website based on voice command"""
        if site_name in self.websites:
            url = self.websites[site_name]
            webbrowser.open(url)
            self.speak(f"Opening {site_name.capitalize()}")
        else:
            self.speak(f"Sorry, I can't open {site_name}.")

    def process_command(self, command):
        """Process voice commands"""
        if not command:
            return
            
        if "open" in command:
            for site in self.websites:
                if site in command:
                    self.open_website(site)
                    return
            self.speak("Which website should I open? Try 'YouTube' or 'Facebook'")
            
        elif any(word in command for word in ["exit", "stop", "bye", "quit", "close", "shut down"]):
            self.speak("Goodbye! Have a great day.")
            self.running = False
            
        elif "help" in command:
            self.speak("You can say: open YouTube, open Facebook, open Instagram, or exit")
            
        else:
            self.speak("I didn't understand. Try saying 'open YouTube' or 'exit'.")

    def run(self):
        """Main execution loop"""
        self.speak("Voice Browser Assistant activated. How can I help you?")
        
        while self.running:
            command = self.listen()
            self.process_command(command)
            time.sleep(1)

        # Clean exit
        self.engine.stop()
        sys.exit(0)

if __name__ == "__main__":
    try:
        assistant = VoiceBrowserAssistant()
        assistant.run()
    except KeyboardInterrupt:
        print("\nProgram stopped by user")
        sys.exit(0)
    except Exception as e:
        print(f"An error occurred: {str(e)}")
        sys.exit(1)
```
