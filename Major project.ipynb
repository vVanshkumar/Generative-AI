```python
import speech_recognition as sr
import pyttsx3
import webbrowser
import datetime
import pywhatkit
import time
import os

# Initialize text-to-speech engine
engine = pyttsx3.init()
voices = engine.getProperty('voices')
engine.setProperty('voice', voices[1].id)  # 0 for male, 1 for female voice
engine.setProperty('rate', 150)  # Speed of speech

def speak(text):
    """Convert text to speech"""
    engine.say(text)
    engine.runAndWait()

def take_command():
    """Take microphone input and return text"""
    r = sr.Recognizer()
    with sr.Microphone() as source:
        print("Listening...")
        r.pause_threshold = 1
        audio = r.listen(source)
        
    try:
        print("Recognizing...")
        query = r.recognize_google(audio, language='en-in')
        print(f"User said: {query}\n")
        
    except Exception as e:
        print("Say that again please...")
        return "None"
    return query.lower()

def greet():
    """Greet the user based on time"""
    hour = datetime.datetime.now().hour
    if 0 <= hour < 12:
        speak("Good Morning! How can I help you?")
    elif 12 <= hour < 18:
        speak("Good Afternoon! How can I help you?")
    else:
        speak("Good Evening! How can I help you?")

def get_time():
    """Get current time and speak it"""
    current_time = datetime.datetime.now().strftime("%I:%M %p")
    speak(f"The current time is {current_time}")

def get_date():
    """Get current date and speak it"""
    current_date = datetime.datetime.now().strftime("%B %d, %Y")
    speak(f"Today's date is {current_date}")

def get_day():
    """Get current day and speak it"""
    current_day = datetime.datetime.now().strftime("%A")
    speak(f"Today is {current_day}")

def open_website(url):
    """Open a website in default browser"""
    webbrowser.open(url)
    speak(f"Opening {url.split('//')[1].split('/')[0]}")

def search_google(query):
    """Search on Google"""
    search_url = f"https://www.google.com/search?q={query.replace(' ', '+')}"
    webbrowser.open(search_url)
    speak(f"Searching Google for {query}")

def search_youtube(query):
    """Search on YouTube"""
    pywhatkit.playonyt(query)
    speak(f"Searching YouTube for {query}")

def search_social_media(platform, query):
    """Search on social media platforms"""
    platforms = {
        'facebook': 'https://www.facebook.com/search/top/?q=',
        'twitter': 'https://twitter.com/search?q=',
        'instagram': 'https://www.instagram.com/explore/tags/',
        'linkedin': 'https://www.linkedin.com/search/results/all/?keywords=',
        'gmail': 'https://mail.google.com/mail/u/0/#search/'
    }
    
    if platform in platforms:
        search_url = f"{platforms[platform]}{query.replace(' ', '+')}"
        webbrowser.open(search_url)
        speak(f"Searching {platform} for {query}")
    else:
        speak("Sorry, I don't support searching on that platform.")

def main():
    greet()
    while True:
        query = take_command().lower()
        
        if 'time' in query:
            get_time()
            
        elif 'date' in query:
            get_date()
            
        elif 'day' in query:
            get_day()
            
        elif 'open google' in query:
            open_website("https://www.google.com")
            
        elif 'open youtube' in query:
            open_website("https://www.youtube.com")
            
        elif 'open gmail' in query:
            open_website("https://mail.google.com")
            
        elif 'open twitter' in query:
            open_website("https://twitter.com")
            
        elif 'open facebook' in query:
            open_website("https://www.facebook.com")
            
        elif 'open instagram' in query:
            open_website("https://www.instagram.com")
            
        elif 'open linkedin' in query:
            open_website("https://www.linkedin.com")
            
        elif 'search on google' in query:
            speak("What would you like me to search on Google?")
            search_query = take_command()
            if search_query != "none":
                search_google(search_query)
                
        elif 'search on youtube' in query:
            speak("What would you like me to search on YouTube?")
            search_query = take_command()
            if search_query != "none":
                search_youtube(search_query)
                
        elif 'search on facebook' in query:
            speak("What would you like me to search on Facebook?")
            search_query = take_command()
            if search_query != "none":
                search_social_media('facebook', search_query)
                
        elif 'search on twitter' in query:
            speak("What would you like me to search on Twitter?")
            search_query = take_command()
            if search_query != "none":
                search_social_media('twitter', search_query)
                
        elif 'search on instagram' in query:
            speak("What would you like me to search on Instagram?")
            search_query = take_command()
            if search_query != "none":
                search_social_media('instagram', search_query)
                
        elif 'search on linkedin' in query:
            speak("What would you like me to search on LinkedIn?")
            search_query = take_command()
            if search_query != "none":
                search_social_media('linkedin', search_query)
                
        elif 'search on gmail' in query:
            speak("What would you like me to search in your Gmail?")
            search_query = take_command()
            if search_query != "none":
                search_social_media('gmail', search_query)
                
        elif 'exit' in query or 'quit' in query or 'stop' in query:
            speak("Goodbye! Have a nice day.")
            break
            
        else:
            speak("I didn't understand that. Can you please repeat?")

if __name__ == "__main__":
    main()
```
