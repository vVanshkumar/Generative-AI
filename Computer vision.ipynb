```python
import cv2
import numpy as np
import ultralytics

print("OpenCV:", cv2.__version__)
print("NumPy:", np.__version__)
print("YOLO:", ultralytics.__version__)
print("Webcam support:", hasattr(cv2, "VideoCapture"))
```

```python

# ==========================================
# PROJECT: REAL-TIME OBJECT DETECTION USING YOLO
# ==========================================

import cv2
from ultralytics import YOLO

# Step 1: Load pretrained YOLO model
print("Loading YOLO model...")

model = YOLO("yolo11n.pt")

print("YOLO model loaded successfully!")

# Step 2: Open webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Webcam not found!")
    raise SystemExit

print("Webcam started successfully!")
print("Press Q to stop detection.")

# Step 3: Real-time object detection
while True:

    ret, frame = cap.read()

    if not ret:
        print("Cannot read webcam frame.")
        break

    # Step 4: Detect objects
    results = model(
        frame,
        conf=0.40,
        imgsz=640,
        verbose=False
    )

    # Step 5: Draw bounding boxes and labels
    result_frame = results[0].plot()

    # Step 6: Display detection results
    cv2.imshow(
        "YOLO Real-Time Object Detection",
        result_frame
    )

    # Step 7: Press Q to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Step 8: Release webcam
cap.release()
cv2.destroyAllWindows()

print("YOLO detection stopped successfully!")
```

```python

# ==========================================
# YOLO REAL-TIME OBJECT COUNTING PROJECT
# ==========================================

import cv2
from ultralytics import YOLO
from collections import Counter

# Step 1: Load YOLO model
print("Loading YOLO model...")

model = YOLO("yolo11n.pt")

print("Model loaded successfully!")

# Step 2: Start webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Webcam not available!")
    raise SystemExit

print("Webcam started!")
print("Press Q to exit.")

# Step 3: Run object detection
while True:

    ret, frame = cap.read()

    if not ret:
        print("Frame not received!")
        break

    # Step 4: Detect objects
    results = model(
        frame,
        conf=0.40,
        verbose=False
    )

    # Step 5: Draw bounding boxes
    output = results[0].plot()

    # Step 6: Count detected objects
    object_counts = Counter()

    if results[0].boxes is not None:

        for box in results[0].boxes:
            class_id = int(box.cls[0])
            object_name = model.names[class_id]

            object_counts[object_name] += 1

    # Step 7: Display object counts
    y_position = 35

    cv2.putText(
        output,
        "YOLO OBJECT COUNTING",
        (10, y_position),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 0),
        2
    )

    y_position += 35

    for name, count in object_counts.items():

        text = f"{name}: {count}"

        cv2.putText(
            output,
            text,
            (10, y_position),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.65,
            (0, 255, 255),
            2
        )

        y_position += 30

    # Step 8: Display total count
    total = sum(object_counts.values())

    cv2.putText(
        output,
        f"Total Objects: {total}",
        (10, y_position + 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 0, 0),
        2
    )

    # Step 9: Show output
    cv2.imshow("YOLO Object Counting", output)

    # Step 10: Exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Step 11: Release resources
cap.release()
cv2.destroyAllWindows()

print("Project completed!")
```
