```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

# Download necessary NLTK data (only first time)
nltk.download('punkt')
nltk.download('stopwords')

# Define positive and negative words (very simple rule-based sentiment)
positive_words = ["good", "great", "excellent", "happy", "awesome", "fantastic", "love"]
negative_words = ["bad", "sad", "terrible", "awful", "hate", "poor", "worst"]

# Preprocessing function
def preprocess(text):
    # Tokenization
    tokens = word_tokenize(text.lower())
    
    # Remove stopwords
    stop_words = set(stopwords.words("english"))
    tokens = [word for word in tokens if word.isalpha() and word not in stop_words]
    
    # Stemming
    ps = PorterStemmer()
    tokens = [ps.stem(word) for word in tokens]
    
    return tokens

# Sentiment Analysis (rule-based)
def sentiment_analysis(text):
    tokens = preprocess(text)
    
    pos_count = sum(1 for word in tokens if word in positive_words)
    neg_count = sum(1 for word in tokens if word in negative_words)
    
    if pos_count > neg_count:
        return "Positive Sentiment 🙂"
    elif neg_count > pos_count:
        return "Negative Sentiment 🙁"
    else:
        return "Neutral 😐"

# Main
if __name__ == "__main__":
    print("Simple NLP Sentiment Analysis Project")
    print("Type 'quit' to exit.\n")
    
    while True:
        user_input = input("Enter a sentence: ")
        if user_input.lower() == "quit":
            print("Goodbye! 👋")
            break
        result = sentiment_analysis(user_input)
        print("Analysis:", result)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from collections import Counter

# Download NLTK data (only needed once)
nltk.download('punkt')
nltk.download('stopwords')

# Function for preprocessing and word frequency
def word_frequency(text):
    # Convert to lowercase
    text = text.lower()
    
    # Tokenize
    tokens = word_tokenize(text)
    
    # Keep only words (remove punctuation/numbers)
    words = [word for word in tokens if word.isalpha()]
    
    # Remove stopwords (common words like "the", "is", etc.)
    stop_words = set(stopwords.words("english"))
    filtered_words = [word for word in words if word not in stop_words]
    
    # Count word frequencies
    word_counts = Counter(filtered_words)
    
    return word_counts

# Main program
if __name__ == "__main__":
    print("Simple NLP Project: Word Frequency Counter")
    text = input("Enter a sentence or paragraph:\n")
    
    freq = word_frequency(text)
    print("\nTop 5 most common words:")
    for word, count in freq.most_common(5):
        print(word, ":", count)
```

```python
import nltk
from nltk.corpus import stopwords
import random

# Download NLTK data
nltk.download('punkt')
nltk.download('stopwords')

# Sample dataset (Normally you would load a larger dataset)
# Label: "spam" or "ham"
dataset = [
    ("Hello, how are you?", "ham"),
    ("Win money now!!!", "spam"),
    ("Call me when you are free", "ham"),
    ("Congratulations, you have won a lottery!", "spam"),
    ("Let's meet for lunch tomorrow", "ham"),
    ("Earn cash quickly, click this link", "spam"),
    ("Are we still meeting today?", "ham"),
    ("You have been selected for a free prize", "spam"),
]

# Preprocessing function
def preprocess(sentence):
    tokens = nltk.word_tokenize(sentence.lower())
    stop_words = set(stopwords.words("english"))
    words = [word for word in tokens if word.isalpha() and word not in stop_words]
    return words

# Build features
all_words = []
for (sentence, label) in dataset:
    all_words.extend(preprocess(sentence))

all_words = nltk.FreqDist(all_words)
word_features = list(all_words.keys())[:200]  # top 200 words as features

def extract_features(document):
    words = set(preprocess(document))
    features = {}
    for word in word_features:
        features[f'contains({word})'] = (word in words)
    return features

# Create feature sets
feature_sets = [(extract_features(sentence), label) for (sentence, label) in dataset]

# Shuffle data
random.shuffle(feature_sets)

# Train-test split
train_size = int(len(feature_sets) * 0.8)
train_set, test_set = feature_sets[:train_size], feature_sets[train_size:]

# Train Naive Bayes Classifier
classifier = nltk.NaiveBayesClassifier.train(train_set)

# Test accuracy
accuracy = nltk.classify.accuracy(classifier, test_set) * 100
print(f"Accuracy: {accuracy:.2f}%")

# Show most informative features
classifier.show_most_informative_features(5)

# Test custom input
while True:
    msg = input("\nEnter a message (or 'quit' to exit): ")
    if msg.lower() == "quit":
        break
    features = extract_features(msg)
    result = classifier.classify(features)
    print("Prediction:", result)
```

```python
# spam_classifier_nltk_debug.py
import nltk
import random
import traceback
from nltk.probability import FreqDist
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

def ensure_nltk_data():
    """Try to ensure required NLTK resources are available."""
    resources = {
        'punkt': 'tokenizers/punkt',
        'stopwords': 'corpora/stopwords'
    }
    for name, path in resources.items():
        try:
            nltk.data.find(path)
        except LookupError:
            print(f"NLTK resource '{name}' not found — downloading...")
            nltk.download(name, quiet=True)
            try:
                nltk.data.find(path)
                print(f"Downloaded '{name}' successfully.")
            except LookupError:
                print(f"Failed to download '{name}'. You may need to run:\n"
                      "    python -m nltk.downloader punkt stopwords")
                raise

def preprocess(text):
    """Lowercase, tokenize, remove non-alpha tokens and stopwords."""
    tokens = word_tokenize(text.lower())
    stop_words = set(stopwords.words("english"))
    return [t for t in tokens if t.isalpha() and t not in stop_words]

def extract_features(document_words_set, word_features):
    return {f'contains({w})': (w in document_words_set) for w in word_features}

def safe_train_test_split(feature_sets, test_ratio=0.2):
    """Return (train_set, test_set) guaranteeing both are non-empty."""
    random.shuffle(feature_sets)
    n_train = int(len(feature_sets) * (1 - test_ratio))
    # ensure at least 1 in train and 1 in test
    n_train = max(1, min(len(feature_sets)-1, n_train))
    train = feature_sets[:n_train]
    test = feature_sets[n_train:]
    return train, test

def main():
    try:
        ensure_nltk_data()
    except Exception:
        print("Please ensure NLTK data is installed and you have internet access for downloads.")
        return

    # Expanded but still small sample dataset for demonstration
    dataset = [
        ("I'm going to the market", "ham"),
        ("Are you coming tonight?", "ham"),
        ("Hey, long time no see!", "ham"),
        ("Don't forget the meeting at 10", "ham"),
        ("Lunch at 1pm?", "ham"),
        ("Can you call me back?", "ham"),
        ("Happy birthday! Wish you all the best", "ham"),
        ("Please send the report by EOD", "ham"),
        ("See you at the event tomorrow", "ham"),
        ("Let's catch up this weekend", "ham"),
        ("Your appointment is confirmed", "ham"),
        ("Reminder: project deadline is Friday", "ham"),
        ("Congratulations! You have won a $1000 gift card. Click here to claim.", "spam"),
        ("URGENT: Your account has been compromised. Reset your password now.", "spam"),
        ("Win cash prizes now!!!", "spam"),
        ("Lowest price on medicines, buy now", "spam"),
        ("You have been selected for a FREE vacation", "spam"),
        ("Earn $5000 per week from home", "spam"),
        ("This is not a scam, claim your prize", "spam"),
        ("Act fast: limited offer on electronics", "spam"),
        ("Click the link to win an iPhone", "spam"),
        ("Get rich quick with this one trick", "spam"),
        ("Exclusive deal: 90% off, buy today", "spam"),
        ("Congratulations, you've been selected for a free trial", "spam"),
    ]

    # Build word features
    all_words = []
    for (sent, label) in dataset:
        all_words.extend(preprocess(sent))
    if not all_words:
        print("No tokens found after preprocessing — check your preprocessing step.")
        return

    freq = FreqDist(all_words)
    word_features = list(freq.keys())[:200]  # top features

    # Build feature sets
    feature_sets = []
    for (sent, label) in dataset:
        words = set(preprocess(sent))
        features = extract_features(words, word_features)
        feature_sets.append((features, label))

    print(f"Total samples: {len(feature_sets)}   (spam: {sum(1 for _,l in feature_sets if l=='spam')}, "
          f"ham: {sum(1 for _,l in feature_sets if l=='ham')})")

    # Safe split
    train_set, test_set = safe_train_test_split(feature_sets, test_ratio=0.25)
    print(f"Train size: {len(train_set)}  Test size: {len(test_set)}")

    # Train classifier
    try:
        classifier = nltk.NaiveBayesClassifier.train(train_set)
    except Exception as e:
        print("Error training classifier:")
        traceback.print_exc()
        return

    # Evaluate
    try:
        accuracy = nltk.classify.accuracy(classifier, test_set) * 100
        print(f"Accuracy: {accuracy:.2f}%")
    except Exception:
        print("Error computing accuracy (maybe test set is empty):")
        traceback.print_exc()

    classifier.show_most_informative_features(10)

    # Interactive classify loop (safe)
    try:
        while True:
            msg = input("\nEnter a message (or 'quit' to exit): ").strip()
            if not msg:
                continue
            if msg.lower() == 'quit':
                break
            feats = extract_features(set(preprocess(msg)), word_features)
            print("Prediction:", classifier.classify(feats))
    except (KeyboardInterrupt, EOFError):
        print("\nExiting...")

if __name__ == "__main__":
    main()
```

```python
import cv2

def face_eye_detection():
    # Load Haar Cascade Classifiers (included in OpenCV)
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
    eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye.xml")

    # Open webcam (0 = default camera)
    cap = cv2.VideoCapture(0)

    print("Press 'q' to quit...")

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Failed to grab frame")
            break

        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # Detect faces
        faces = face_cascade.detectMultiScale(gray, scaleFactor=1.3, minNeighbors=5)

        for (x, y, w, h) in faces:
            # Draw rectangle around face
            cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)

            roi_gray = gray[y:y + h, x:x + w]
            roi_color = frame[y:y + h, x:x + w]

            # Detect eyes within face ROI
            eyes = eye_cascade.detectMultiScale(roi_gray)
            for (ex, ey, ew, eh) in eyes:
                cv2.rectangle(roi_color, (ex, ey), (ex + ew, ey + eh), (0, 255, 0), 2)

        # Show output
        cv2.imshow("Face and Eye Detection", frame)

        # Exit on 'q' key
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    face_eye_detection()
```

```python
import cv2

def face_eye_smile_detection():
    # Load Haar Cascade classifiers
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
    eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye.xml")
    smile_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_smile.xml")

    # Start webcam
    cap = cv2.VideoCapture(0)
    print("Press 'q' to quit...")

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Failed to grab frame")
            break

        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # Detect faces
        faces = face_cascade.detectMultiScale(gray, scaleFactor=1.3, minNeighbors=5)

        for (x, y, w, h) in faces:
            # Draw rectangle around face
            cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)

            roi_gray = gray[y:y + h, x:x + w]
            roi_color = frame[y:y + h, x:x + w]

            # Detect eyes within face ROI
            eyes = eye_cascade.detectMultiScale(roi_gray, scaleFactor=1.1, minNeighbors=10)
            for (ex, ey, ew, eh) in eyes:
                cv2.rectangle(roi_color, (ex, ey), (ex + ew, ey + eh), (0, 255, 0), 2)

            # Detect smiles within face ROI
            smiles = smile_cascade.detectMultiScale(roi_gray, scaleFactor=1.7, minNeighbors=22)
            for (sx, sy, sw, sh) in smiles:
                cv2.rectangle(roi_color, (sx, sy), (sx + sw, sy + sh), (0, 0, 255), 2)

        # Display the output
        cv2.imshow("Face, Eye, and Smile Detection", frame)

        # Quit on 'q' key
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    face_eye_smile_detection()
```
