```python

import cv2

# Load pre-trained detection models
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

eye_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_eye.xml"
)

smile_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_smile.xml"
)

# Start webcam
cap = cv2.VideoCapture(0)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # Convert image into grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = face_cascade.detectMultiScale(
        gray, 1.3, 5
    )

    for (x, y, w, h) in faces:

        # Draw face rectangle
        cv2.rectangle(
            frame, (x, y), (x+w, y+h),
            (0, 255, 0), 2
        )

        cv2.putText(
            frame, "Face Detected",
            (x, y-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7, (0, 255, 0), 2
        )

        # Select face region
        roi_gray = gray[y:y+h, x:x+w]
        roi_color = frame[y:y+h, x:x+w]

        # Detect eyes
        eyes = eye_cascade.detectMultiScale(
            roi_gray, 1.1, 5
        )

        for (ex, ey, ew, eh) in eyes:

            cv2.rectangle(
                roi_color,
                (ex, ey), (ex+ew, ey+eh),
                (255, 0, 0), 2
            )

        # Detect smile in lower face
        lower_face = roi_gray[h//2:h, :]

        smiles = smile_cascade.detectMultiScale(
            lower_face, 1.7, 20
        )

        for (sx, sy, sw, sh) in smiles:

            cv2.rectangle(
                roi_color,
                (sx, sy+h//2),
                (sx+sw, sy+sh+h//2),
                (0, 255, 255), 2
            )

            cv2.putText(
                frame, "Smile",
                (x+sx, y+sy+h//2-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7, (0, 255, 255), 2
            )

    # Display output
    cv2.imshow("AI Detection System", frame)

    # Press Q to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python

# ==========================================================
# YOLO BASED REAL-TIME AI DETECTION PROJECT
# Face + Eye + Smile + Hand Gesture + Emotion
# ==========================================================

import cv2
import numpy as np
import os
import urllib.request
from ultralytics import YOLO

# ==========================================================
# 1. LOAD YOLO MODEL
# ==========================================================

print("Loading YOLO model...")

model = YOLO("yolov8n.pt")

print("YOLO model loaded successfully!")

# ==========================================================
# 2. LOAD OPENCV FACE, EYE AND SMILE MODELS
# ==========================================================

face_model = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

eye_model = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_eye.xml"
)

smile_model = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_smile.xml"
)

# ==========================================================
# 3. LOAD MEDIAPIPE FOR HAND GESTURES
# ==========================================================

import mediapipe as mp

mp_hands = mp.solutions.hands

hands = mp_hands.Hands(
    static_image_mode=False,
    max_num_hands=2,
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5
)

# ==========================================================
# 4. LOAD DEEPFACE FOR EMOTION DETECTION
# ==========================================================

from deepface import DeepFace

print("All detection modules loaded!")

# ==========================================================
# 5. START WEBCAM
# ==========================================================

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("ERROR: Webcam not found!")
    raise SystemExit

print("AI Detection Started!")
print("Press Q to stop.")

# ==========================================================
# 6. MAIN DETECTION LOOP
# ==========================================================

emotion_counter = 0
current_emotion = "Detecting..."

while True:

    ret, frame = cap.read()

    if not ret:
        print("Cannot read webcam!")
        break

    frame = cv2.flip(frame, 1)

    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    # ======================================================
    # YOLO OBJECT DETECTION
    # ======================================================

    results = model(
        frame,
        conf=0.4,
        verbose=False
    )

    for box in results[0].boxes:

        class_id = int(box.cls[0])

        object_name = model.names[class_id]

        x1, y1, x2, y2 = map(
            int, box.xyxy[0]
        )

        cv2.rectangle(
            frame,
            (x1, y1),
            (x2, y2),
            (255, 0, 0),
            2
        )

        cv2.putText(
            frame,
            "YOLO: " + object_name,
            (x1, max(20, y1 - 10)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255, 0, 0),
            2
        )

    # ======================================================
    # FACE DETECTION
    # ======================================================

    faces = face_model.detectMultiScale(
        gray,
        scaleFactor=1.3,
        minNeighbors=5
    )

    for (x, y, w, h) in faces:

        cv2.rectangle(
            frame,
            (x, y),
            (x+w, y+h),
            (0, 255, 0),
            2
        )

        cv2.putText(
            frame,
            "FACE",
            (x, y-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )

        roi_gray = gray[y:y+h, x:x+w]

        # ==================================================
        # EYE DETECTION
        # ==================================================

        eyes = eye_model.detectMultiScale(
            roi_gray,
            scaleFactor=1.1,
            minNeighbors=5
        )

        for (ex, ey, ew, eh) in eyes:

            cv2.rectangle(
                frame,
                (x+ex, y+ey),
                (x+ex+ew, y+ey+eh),
                (255, 0, 0),
                2
            )

            cv2.putText(
                frame,
                "EYE",
                (x+ex, y+ey-5),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (255, 0, 0),
                1
            )

        # ==================================================
        # SMILE DETECTION
        # ==================================================

        lower_face = roi_gray[h//2:h, :]

        smiles = smile_model.detectMultiScale(
            lower_face,
            scaleFactor=1.7,
            minNeighbors=20
        )

        for (sx, sy, sw, sh) in smiles:

            cv2.rectangle(
                frame,
                (x+sx, y+h//2+sy),
                (x+sx+sw, y+h//2+sy+sh),
                (0, 255, 255),
                2
            )

            cv2.putText(
                frame,
                "SMILE",
                (x+sx, y+h//2+sy-5),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 255),
                2
            )

        # ==================================================
        # EMOTION DETECTION
        # ==================================================

        emotion_counter += 1

        if emotion_counter >= 20:

            try:

                face_crop = frame[
                    y:y+h,
                    x:x+w
                ]

                if face_crop.size > 0:

                    analysis = DeepFace.analyze(
                        face_crop,
                        actions=["emotion"],
                        enforce_detection=False,
                        detector_backend="skip",
                        silent=True
                    )

                    current_emotion = analysis[0][
                        "dominant_emotion"
                    ]

            except Exception:
                current_emotion = "Unknown"

            emotion_counter = 0

        cv2.putText(
            frame,
            "Emotion: " + current_emotion,
            (x, y+h+25),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 255),
            2
        )

    # ======================================================
    # HAND GESTURE DETECTION
    # ======================================================

    rgb_frame = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2RGB
    )

    hand_results = hands.process(rgb_frame)

    if hand_results.multi_hand_landmarks:

        for hand_landmarks in hand_results.multi_hand_landmarks:

            points = hand_landmarks.landmark

            # Finger tip and joint indexes
            finger_tips = [8, 12, 16, 20]
            finger_joints = [6, 10, 14, 18]

            fingers_up = 0

            for tip, joint in zip(
                finger_tips,
                finger_joints
            ):

                if points[tip].y < points[joint].y:
                    fingers_up += 1

            if fingers_up == 4:
                gesture = "OPEN PALM"

            elif fingers_up == 0:
                gesture = "FIST"

            elif fingers_up == 1:
                gesture = "ONE FINGER"

            elif fingers_up == 2:
                gesture = "PEACE / TWO FINGERS"

            else:
                gesture = "HAND GESTURE"

            height, width, _ = frame.shape

            wrist = points[0]

            wrist_x = int(wrist.x * width)
            wrist_y = int(wrist.y * height)

            cv2.putText(
                frame,
                gesture,
                (wrist_x, wrist_y-20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 255, 0),
                2
            )

            # Draw hand landmarks
            for point in points:

                px = int(point.x * width)
                py = int(point.y * height)

                cv2.circle(
                    frame,
                    (px, py),
                    4,
                    (0, 0, 255),
                    -1
                )

    # ======================================================
    # DISPLAY PROJECT TITLE
    # ======================================================

    cv2.putText(
        frame,
        "YOLO AI DETECTION SYSTEM",
        (20, 35),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 255),
        2
    )

    # ======================================================
    # SHOW OUTPUT
    # ======================================================

    cv2.imshow(
        "AI Face, Eye, Smile, Emotion and Gesture Detection",
        frame
    )

    # Press Q to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# ==========================================================
# 7. RELEASE CAMERA AND CLOSE WINDOWS
# ==========================================================

cap.release()
hands.close()
cv2.destroyAllWindows()

print("AI Detection Stopped Successfully!")
```
