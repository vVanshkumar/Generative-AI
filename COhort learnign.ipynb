```python
# Import Pandas
import pandas as pd

# 1. Creating a DataFrame

# Define data as a dictionary
data = {
    'Name': ['John', 'Sara', 'Mike', 'Emma'],
    'Age': [28, 22, 32, 29],
    'City': ['New York', 'Paris', 'London', 'Tokyo'],
    'Salary': [70000, 65000, 80000, 75000]
}

# Create DataFrame from the dictionary
df = pd.DataFrame(data)

# 2. Display DataFrame

print("DataFrame:")
print(df)

# 3. Basic Operations

# Display column names
print("\nColumn names:")
print(df.columns)

# Access a specific column (e.g., 'Name')
print("\nNames column:")
print(df['Name'])

# Add a new column (e.g., 'Bonus')
df['Bonus'] = df['Salary'] * 0.10
print("\nDataFrame with Bonus column added:")
print(df)

# 4. Filter rows based on a condition (e.g., Age > 25)
filtered_df = df[df['Age'] > 25]
print("\nFiltered DataFrame (Age > 25):")
print(filtered_df)

# 5. Sorting

# Sort DataFrame by Salary in descending order
df_sorted = df.sort_values(by='Salary', ascending=False)
print("\nDataFrame sorted by Salary (descending):")
print(df_sorted)

# 6. Export the DataFrame to CSV

df.to_csv('employees.csv', index=False)
print("\nDataFrame exported to employees.csv")


```
