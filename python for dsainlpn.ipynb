```python
# Example 1: Iterating through a list
fruits = ["apple", "banana", "cherry"]
print("Fruits in the list:")
for fruit in fruits:
    print(fruit)

# Example 2: Iterating through a range of numbers
print("\nNumbers from 1 to 5:")
for num in range(1, 6):  # range(start, end)
    print(num)

# Example 3: Iterating through a string
message = "Python"
print("\nCharacters in the string:")
for char in message:
    print(char)

# Example 4: Using a for loop with conditions
numbers = [1, 2, 3, 4, 5]
print("\nEven numbers in the list:")
for number in numbers:
    if number % 2 == 0:
        print(number)

# Example 5: Nested for loop
print("\nNested for loop example:")
for i in range(1, 4):  # Outer loop
    for j in range(1, 4):  # Inner loop
        print(f"i={i}, j={j}")

# Example 6: Using the break statement
print("\nUsing break in a for loop:")
for num in range(1, 10):
    if num == 5:
        print("Breaking the loop at", num)
        break
    print(num)

# Example 7: Using the continue statement
print("\nUsing continue in a for loop:")
for num in range(1, 6):
    if num == 3:
        print("Skipping", num)
        continue
    print(num)
```

```python
# Import Pandas library
import pandas as pd

# -------------------------------
# 1. Creating a DataFrame
# -------------------------------
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
df = pd.DataFrame(data)
print("DataFrame:")
print(df)

# -------------------------------
# 2. Reading data from a CSV file
# -------------------------------
# Uncomment and replace 'file.csv' with the path to your file
# df_csv = pd.read_csv('file.csv')
# print("\nData from CSV:")
# print(df_csv.head())  # Display first 5 rows

# -------------------------------
# 3. Data Inspection
# -------------------------------
print("\nShape of DataFrame:", df.shape)  # Rows and columns
print("\nColumn names:", df.columns)  # Column names
print("\nData types:\n", df.dtypes)  # Data types of each column
print("\nSummary statistics:\n", df.describe())  # Summary statistics for numeric columns

# -------------------------------
# 4. Data Selection
# -------------------------------
print("\nSelect a column (Name):")
print(df['Name'])  # Select a single column

print("\nSelect multiple columns (Name and Age):")
print(df[['Name', 'Age']])  # Select multiple columns

print("\nSelect rows (using loc):")
print(df.loc[0])  # Select row by index

print("\nSelect rows and columns (using iloc):")
print(df.iloc[1:3, 0:2])  # Select rows 1-2 and columns 0-1

# -------------------------------
# 5. Filtering Data
# -------------------------------
print("\nFilter rows where Age > 25:")
print(df[df['Age'] > 25])

# -------------------------------
# 6. Adding and Dropping Columns
# -------------------------------
df['Salary'] = [50000, 60000, 70000]  # Add a new column
print("\nDataFrame after adding Salary column:")
print(df)

df.drop('City', axis=1, inplace=True)  # Drop the 'City' column
print("\nDataFrame after dropping City column:")
print(df)

# -------------------------------
# 7. Handling Missing Data
# -------------------------------
# Creating a new DataFrame with missing values
data_with_nan = {
    'Name': ['Alice', 'Bob', None],
    'Age': [25, None, 35],
    'City': ['New York', None, 'Chicago']
}
df_nan = pd.DataFrame(data_with_nan)
print("\nDataFrame with Missing Data:")
print(df_nan)

# Filling missing values
df_nan_filled = df_nan.fillna({'Age': df_nan['Age'].mean(), 'City': 'Unknown', 'Name': 'Unknown'})
print("\nDataFrame after filling missing values:")
print(df_nan_filled)

# Dropping rows with missing values
df_nan_dropped = df_nan.dropna()
print("\nDataFrame after dropping rows with missing values:")
print(df_nan_dropped)

# -------------------------------
# 8. Grouping and Aggregation
# -------------------------------
print("\nGroup by Age and calculate mean Salary:")
print(df.groupby('Age')['Salary'].mean())

# -------------------------------
# 9. Exporting Data
# -------------------------------
# Uncomment and replace 'output.csv' with your desired file path
# df.to_csv('output.csv', index=False)
# print("\nDataFrame exported to output.csv")

# -------------------------------
# 10. Visualization with Pandas
# -------------------------------
import matplotlib.pyplot as plt

df['Age'].plot(kind='bar', title='Age Distribution')
plt.xlabel('Index')
plt.ylabel('Age')
plt.show()
```
