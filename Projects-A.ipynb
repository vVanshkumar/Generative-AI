```python
import speech_recognition as sr
import pyttsx3
import webbrowser
import time
import sys
import os

class VoiceAssistant:
    def __init__(self):
        self.recognizer = sr.Recognizer()
        self.engine = pyttsx3.init()
        self.running = True
        self.setup_voice()
        self.websites = {
            "youtube": "https://www.youtube.com",
            "facebook": "https://www.facebook.com",
            "instagram": "https://www.instagram.com",
            "google": "https://www.google.com",
            "wikipedia": "https://www.wikipedia.org",
            "twitter": "https://twitter.com"
        }

    def setup_voice(self):
        """Configure voice settings"""
        voices = self.engine.getProperty('voices')
        self.engine.setProperty('voice', voices[1].id)  # 0 for male, 1 for female
        self.engine.setProperty('rate', 150)

    def speak(self, text):
        """Convert text to speech"""
        print(f"Assistant: {text}")
        self.engine.say(text)
        self.engine.runAndWait()

    def listen(self):
        """Listen to microphone input"""
        with sr.Microphone() as source:
            print("\nListening... (Say 'open YouTube', 'exit', etc.)")
            self.recognizer.adjust_for_ambient_noise(source)
            try:
                audio = self.recognizer.listen(source, timeout=5)
                command = self.recognizer.recognize_google(audio).lower()
                print(f"You said: {command}")
                return command
            except sr.WaitTimeoutError:
                return ""
            except sr.UnknownValueError:
                self.speak("Sorry, I didn't understand that.")
                return ""
            except sr.RequestError:
                self.speak("Sorry, my speech service is down.")
                return ""
            except Exception as e:
                self.speak(f"Error: {str(e)}")
                return ""

    def open_website(self, site_name):
        """Open website in default browser"""
        if site_name in self.websites:
            webbrowser.open(self.websites[site_name])
            self.speak(f"Opening {site_name.capitalize()}")
        else:
            self.speak(f"Sorry, I can't open {site_name}")

    def process_command(self, command):
        """Process voice commands"""
        if not command:
            return
            
        if "open" in command:
            for site in self.websites:
                if site in command:
                    self.open_website(site)
                    return
            self.speak("Which website should I open? Try 'YouTube' or 'Google'")
            
        elif any(word in command for word in ["exit", "stop", "quit", "bye", "close"]):
            self.speak("Goodbye! Have a great day.")
            self.running = False
            
        elif "help" in command:
            sites = ", ".join(self.websites.keys())
            self.speak(f"I can open: {sites}. Just say 'open' followed by the site name.")
            
        else:
            self.speak("I didn't understand. Try saying 'help' for options.")

    def run(self):
        """Main execution loop"""
        self.speak("Voice Assistant ready. How can I help you?")
        
        while self.running:
            command = self.listen()
            self.process_command(command)
            time.sleep(0.5)

        # Clean exit
        self.engine.stop()
        if '__IPYTHON__' not in globals():  # Only exit if not in Jupyter
            sys.exit(0)

if __name__ == "__main__":
    try:
        assistant = VoiceAssistant()
        assistant.run()
    except KeyboardInterrupt:
        print("\nProgram stopped by user")
        sys.exit(0)
    except Exception as e:
        print(f"Error: {str(e)}")
        sys.exit(1)
```

```python
import speech_recognition as sr
import pyttsx3
import datetime
import time

# Initialize the speech engine
engine = pyttsx3.init()
recognizer = sr.Recognizer()
microphone = sr.Microphone()

def speak(text):
    """Convert text to speech"""
    engine.say(text)
    engine.runAndWait()

def get_audio():
    """Listen for audio input and convert to text"""
    with microphone as source:
        print("Listening...")
        recognizer.adjust_for_ambient_noise(source)
        audio = recognizer.listen(source)
        
    try:
        print("Recognizing...")
        query = recognizer.recognize_google(audio, language='en-in')
        print(f"User said: {query}\n")
        return query.lower()
    except Exception as e:
        print("Sorry, I didn't catch that. Could you please repeat?")
        speak("Sorry, I didn't catch that. Could you please repeat?")
        return None

def get_time():
    """Get current time in 12-hour format"""
    now = datetime.datetime.now()
    current_time = now.strftime("%I:%M %p")
    return f"The current time is {current_time}"

def get_date():
    """Get current date"""
    now = datetime.datetime.now()
    current_date = now.strftime("%A, %B %d, %Y")
    return f"Today is {current_date}"

def main():
    speak("Hello! I'm your voice assistant. How can I help you today?")
    
    while True:
        query = get_audio()
        
        if query:
            if 'time' in query:
                time_response = get_time()
                print(time_response)
                speak(time_response)
            elif 'date' in query:
                date_response = get_date()
                print(date_response)
                speak(date_response)
            elif 'exit' in query or 'quit' in query or 'stop' in query:
                speak("Goodbye! Have a nice day.")
                print("Exiting the program...")
                break
            else:
                response = "I can only tell you the time or date. Try asking about those."
                print(response)
                speak(response)

if __name__ == "__main__":
    main()
```
