```python
import cv2
import matplotlib.pyplot as plt
import urllib.request
import os

# -----------------------------
# Download Haar Cascade XML
# -----------------------------
cascade_url = "https://raw.githubusercontent.com/opencv/opencv/master/data/haarcascades/haarcascade_frontalface_default.xml"
cascade_path = "haarcascade_frontalface.xml"

if not os.path.exists(cascade_path):
    urllib.request.urlretrieve(cascade_url, cascade_path)

# -----------------------------
# Download a sample image
# -----------------------------
image_url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
image_path = "face.jpg"

if not os.path.exists(image_path):
    urllib.request.urlretrieve(image_url, image_path)

# -----------------------------
# Load cascade and image
# -----------------------------
face_cascade = cv2.CascadeClassifier(cascade_path)
img = cv2.imread(image_path)

# Safety check
if img is None:
    raise FileNotFoundError("Image not loaded")

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# -----------------------------
# Face Detection
# -----------------------------
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.3,
    minNeighbors=5
)

# Draw rectangles
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x + w, y + h), (255, 0, 0), 2)

# Convert BGR to RGB for Jupyter display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Show result
plt.imshow(img_rgb)
plt.axis("off")
```

```python
import cv2

# Load Haar Cascade files
face_cascade = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier("haarcascade_eye.xml")

# Start webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Convert to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.3,
        minNeighbors=5
    )

    # Draw rectangle around faces
    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x+w, y+h), (255, 0, 0), 2)

        # Region of Interest (ROI) for eyes
        roi_gray = gray[y:y+h, x:x+w]
        roi_color = frame[y:y+h, x:x+w]

        # Detect eyes
        eyes = eye_cascade.detectMultiScale(
            roi_gray,
            scaleFactor=1.1,
            minNeighbors=5
        )

        # Draw rectangle around eyes
        for (ex, ey, ew, eh) in eyes:
            cv2.rectangle(roi_color, (ex, ey), (ex+ew, ey+eh), (0, 255, 0), 2)

    # Show output
    cv2.imshow("Face and Eye Detection", frame)

    # Press 'q' to quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()
```

```python
# ==============================
# 1. Import Required Libraries
# ==============================
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
import matplotlib.pyplot as plt
import numpy as np

# ==============================
# 2. Load Dataset (MNIST)
# ==============================
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# ==============================
# 3. Data Preprocessing
# ==============================
# Normalize
X_train = X_train / 255.0
X_test = X_test / 255.0

# Reshape for CNN
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

# ==============================
# 4. Build CNN Model
# ==============================
model = Sequential()

model.add(Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)))
model.add(MaxPooling2D((2,2)))

model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D((2,2)))

model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dense(10, activation='softmax'))

# ==============================
# 5. Compile Model
# ==============================
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# ==============================
# 6. Train Model
# ==============================
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    validation_split=0.1
)

# ==============================
# 7. Evaluate Model
# ==============================
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Accuracy:", accuracy)

# ==============================
# 8. Prediction on Test Image
# ==============================
predictions = model.predict(X_test)

index = 0
plt.imshow(X_test[index].reshape(28,28), cmap='gray')
plt.title(f"Predicted Digit: {np.argmax(predictions[index])}")
plt.axis('off')
plt.show()

# ==============================
# 9. Save Model
# ==============================
model.save("mnist_cnn_model.h5")
print("Model saved successfully!")
```

```python
# ==============================
# 1. Import Required Libraries
# ==============================
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
import matplotlib.pyplot as plt
import numpy as np

# ==============================
# 2. Load CIFAR-10 Dataset
# ==============================
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()

class_names = ['airplane','car','bird','cat','deer',
               'dog','frog','horse','ship','truck']

# ==============================
# 3. Data Preprocessing
# ==============================
# Normalize pixel values
X_train = X_train / 255.0
X_test = X_test / 255.0

# ==============================
# 4. Build CNN Model
# ==============================
model = Sequential()

model.add(Conv2D(32, (3,3), activation='relu', input_shape=(32,32,3)))
model.add(MaxPooling2D((2,2)))

model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D((2,2)))

model.add(Conv2D(128, (3,3), activation='relu'))
model.add(MaxPooling2D((2,2)))

model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(10, activation='softmax'))

# ==============================
# 5. Compile Model
# ==============================
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# ==============================
# 6. Train Model
# ==============================
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    validation_split=0.1,
    batch_size=64
)

# ==============================
# 7. Evaluate Model
# ==============================
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Accuracy:", accuracy)

# ==============================
# 8. Prediction on Test Image
# ==============================
predictions = model.predict(X_test)

index = 15
plt.imshow(X_test[index])
plt.title(f"Predicted: {class_names[np.argmax(predictions[index])]}")
plt.axis('off')
plt.show()

# ==============================
# 9. Save Model
# ==============================
model.save("cifar10_cnn_model.h5")
print("Model saved successfully!")
```
