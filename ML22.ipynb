```python
import pandas
```

```python
df = pandas.read_csv('salary.csv')
```

```python
x = df["Year of Experience"]
```

```python
y = df["Salary"]
```

```python
x.ndim
```

```python
X = x.values.reshape(-1,1)
```

```python
X.ndim
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X,y)
```

```python
model.coef_
```

```python
model.predict([[5.3]])
```

```python
#mx = y
```

```python
model.intercept_
```
